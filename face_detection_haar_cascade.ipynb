```python
# Some imports:
import os
import sys
import numpy as np
import cv2 as cv
from src import utils
```

```python
root_dir_folder = "D:\\DEV\\INTECH\\casia_dataset\\casia_iris_distance"
```

```python
# extracting subfolders where images are stored:

list_dir = os.listdir(root_dir_folder)
subfolder_users = [elt for elt in list_dir if os.path.isdir(os.path.join(root_dir_folder,elt))]
```

```python
print(subfolder_users)
```

```python
subfolder_subusers = ['043', '044']
```

```python
# initialize a haar cascade face detector:
#face_cascade = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_frontalface_default.xml')
face_cascade_alt = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_frontalface_alt.xml')
face_cascade_alt2 = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_frontalface_alt2.xml')

# initialize a haar cascade eye detector:
eye_cascade = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_eye.xml')
```

```python
image_path_folders = []

for folder in subfolder_subusers:
    print("Operating on user:", folder)

    user_image_path = os.path.join(root_dir_folder, folder)

    for img_file in os.listdir(user_image_path):

        if not img_file.lower().endswith('.jpg'):
            continue

        img_path = os.path.join(user_image_path, img_file)
        image_path_folders.append(img_path)

        img = cv.imread(img_path)
        if img is None:
            print("Failed to read:", img_path)
            continue

        gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
        gradient_img = utils.preprocess_gradient(gray_img)

        img_name = os.path.splitext(img_file)[0]

        # -------------------------------------------------
        # ALT1 FACE DETECTOR
        # -------------------------------------------------
        print("Detecting face (alt1):", img_name)

        alt1_base_dir = os.path.join(user_image_path, img_name, 'alt1_detector')
        iris_dir1 = os.path.join(alt1_base_dir, 'irises')

        faces_alt1 = utils.detect_faces(
            gradient_img,
            face_cascade_alt,
            gray_img,
            alt1_base_dir,
            img_name
        )

        irises1 = []
        if len(faces_alt1) > 0:
            irises1= utils.detect_save_iris2(faces_alt1, eye_cascade, gray_img, img_name, iris_dir1)

        # -------------------------------------------------
        # ALT2 FACE DETECTOR
        # -------------------------------------------------
        print("Detecting face (alt2):", img_name)

        alt2_base_dir = os.path.join(user_image_path, img_name, 'alt2_detector')
        iris_dir2 = os.path.join(alt2_base_dir, 'irises')

        faces_alt2 = utils.detect_faces(
            gradient_img,
            face_cascade_alt2,
            gray_img,
            alt2_base_dir,
            img_name
        )

        irises2 = []
        if len(faces_alt2) > 0:
            irises2 = utils.detect_save_iris2(faces_alt2, eye_cascade,
                                                        gray_img, 
                                                        img_name, 
                                                        iris_dir2)

        # -------------------------------------------------
        # Logging
        # -------------------------------------------------
        if len(faces_alt1) == 0 and len(faces_alt2) == 0:
            print("No face detected:", img_name)
        else:
            #print(f"Irises detected - alt1: {len(irises1)}, alt2: {len(irises2)}")
            print("Irises detected with alt1 and alt2")

    # Debug stop
    #if folder == '031':
        #break
```
